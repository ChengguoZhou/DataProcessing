## 题目3：基于已有 Series 重新取值

## 题目5：标签索引与位置索引

已知：

```python
s = pd.Series([15, 25, 35, 45], index=['w', 'x', 'y', 'z'])
```

分别使用 'loc'、'iloc'、'at'、'iat' 取出值 '25'，并打印结果。

````python
# 新建序列s
s = pd.Series([15, 25, 35, 45], index=['w', 'x', 'y', 'z'])
# 分别使用 'loc(显式索引)'、'iloc（隐式索引，默认从0开始）'、'at'、'iat' 取出值 '25'
print(s.loc['x'])
print(s.iloc[1])
print(s.at['x'])
print(s.iat[1])
````

## 题目6：基本访问与布尔筛选

已知：

'''python
s = pd.Series([72, 88, 95, 61, 79], index=['张三', '李四', '王五', '赵六', '钱七'])
'''

完成以下要求：

1. 取出 '王五' 的成绩。
2. 取出前 3 个元素。
3. 筛选出成绩大于等于 80 的学生。

````python
# 创建序列s
s = pd.Series([72, 88, 95, 61, 79], index=['张三', '李四', '王五', '赵六', '钱七'])
# 取出“王五”的成绩
print(s.loc['王五'])
# 取出前三个元素
print(s.iloc[0:3])
# 筛选出成绩>=80的学生
print(s.loc[s >= 80])
````

## 题目8：缺失值检测与非缺失统计

创建 Series：'[10, np.nan, 30, None, 50, 60]'，索引设为 '['A', 'B', 'C', 'D', 'E', 'F']'。

完成以下要求：

1. 打印该 Series。

## 题目11：常用统计函数

对下面的 Series 分别计算：平均值、总和、标准差、方差、最小值、最大值、中位数。

```python
s = pd.Series([12, 18, 25, 30, 22, 27])
```

````python
# 创建序列s
s = pd.Series([12, 18, 25, 30, 22, 27])
# 输出平均值、总和、标准差、方差、最小值、最大值、中位数
print(f"s的平均值为: {s.mean(): .2f}")
print(f"s的总和为: {s.sum()}")
print(f"s的标准差为: {s.std(): .2f}")
print(f"s的方差为: {s.var(): .2f}")
print(f"s的最大值为: {s.max()}")
print(f"s的最小值为: {s.min()}")
print(f"s的中位数值为: {s.median()}")
````

## 题目12：分位数与众数

已知：

'''python
s = pd.Series([10, 15, 15, 20, 25, 30, 30, 30])
'''

1. 计算该 Series 的 0.75 分位数。
2. 计算该 Series 的众数。

````python
# 新建序列s
s = pd.Series([10, 15, 15, 20, 25, 30, 30, 30])
# 计算0.75分位数
print("s的0.75分位数: ", s.quantile(0.75))
# 计算众数
print("s的众数为: ", *s.mode().values)
````

## 题目13：频数统计、去重与唯一值数量

已知：

'''python
s = pd.Series(['A', 'B', 'A', 'C', 'B', 'A', 'D'])
'''

完成以下要求：

1. 统计每个唯一值出现的次数。
2. 去除重复值并查看结果。
3. 获取去重后的值数组。
4. 统计唯一值个数。

````python
# 新建序列s
s = pd.Series(['A', 'B', 'A', 'C', 'B', 'A', 'D'])
# 统计每个唯一值出现的次数
print(s.value_counts())
# 去除重复值并查看结果
print(s.drop_duplicates())
# 获取去重后的值数组
uniq_s = s.unique()
print(uniq_s)
# 统计唯一值个数
print(len(uniq_s))
````

## 题目14：按值排序与按索引排序

已知：

'''python
s = pd.Series([88, 72, 95, 81], index=['b', 'd', 'a', 'c'])
'''

1. 按值升序排序。
2. 按索引升序排序。

````python
# 新建序列s
s = pd.Series([88, 72, 95, 81], index=['b', 'd', 'a', 'c'])
# 按值升序排序
print(s.sort_values(ascending=True))
# 按索引升序排序
print(s.sort_index(ascending=True))
````

## 题目15：replace 与 sample

已知：

'''python
s = pd.Series(['一线', '二线', '三线', '二线', '一线', '四线'])
'''

完成以下要求：

1. 使用 'replace()' 将 '二线' 替换为 '新一线'。
2. 使用 'sample()' 随机抽取 3 个元素。

````python
# 新建序列s
s = pd.Series(['一线', '二线', '三线', '二线', '一线', '四线'])
# 使用replace()将“二线”替换为“新一线”
s = s.replace('二线', '新一线')
print(s)
# 使用sample()随机抽取3个元素
print(s.sample(3))
````
