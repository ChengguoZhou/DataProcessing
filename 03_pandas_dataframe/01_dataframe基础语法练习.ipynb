## 题目1：使用字典创建 DataFrame

创建一个名为 `students` 的 DataFrame，包含以下数据：

```python
{
    'name': ['Tom', 'Jack', 'Alice', 'Bob'],
    'age': [18, 20, 19, 21],
    'score': [86.5, 92.0, 88.5, 79.0]
}
```

要求：

## 题目5：loc 与 iloc 访问行和列

已知：

```python
df = pd.DataFrame({
    'name': ['Tom', 'Jack', 'Alice', 'Bob'],
    'age': [15, 17, 20, 26],
    'score': [60.5, 80.0, 30.6, 70.0]
}, index=[1, 2, 3, 4])
```

完成以下要求：

1. 使用 `loc` 和 `iloc` 分别取出第 4 行。
2. 使用 `loc` 和 `iloc` 分别取出 `name` 列。
3. 使用 `at`、`iat`、`loc`、`iloc` 四种方式分别取出 `Alice` 的 `score`。

````python
df = pd.DataFrame({
    'name': ['Tom', 'Jack', 'Alice', 'Bob'],
    'age': [15, 17, 20, 26],
    'score': [60.5, 80.0, 30.6, 70.0]
}, index=[1, 2, 3, 4])
print(df)
# 使用 `loc` 和 `iloc` 分别取出第 4 行。
print("取出第4行")
print(df.loc[4])
print(df.iloc[3])
# 使用 `loc` 和 `iloc` 分别取出 `name` 列。
print("取出name列")
print(df.loc[:, 'name'])
print(df.iloc[:, 0])
# 使用 `at`、`iat`、`loc`、`iloc` 四种方式分别取出 `Alice` 的 `score#
print("取出Alice的score")
print(df.at[3, 'score'])
print(df.iat[2, 2])
print(df.loc[3, 'score'])
print(df.iloc[2, 2])
````

## 题目9：isin 与 isna

创建如下 DataFrame：

```python

    'name': ['Tom', 'Jack', None, 'Bob'],

    'city': ['北京', '上海', '广州', '北京']
```

## 题目10：常用统计函数

已知：

```python
df = pd.DataFrame({
    'math': [85, 92, 78, 88, 95],
    'english': [90, 88, 85, 92, 80],
    'physics': [75, 80, 88, 85, 90]
})
```

分别计算：

1. `math` 列总和。
2. `english` 列最大值。
3. `physics` 列最小值。
4. `math` 列平均值。
5. `english` 列中位数。
6. `physics` 列众数。
7. `math` 列标准差、方差、0.25 分位数。

````python
# 创建DataFrame
df = pd.DataFrame({
    'math': [85, 92, 78, 88, 95],
    'english': [90, 88, 85, 92, 80],
    'physics': [75, 80, 88, 85, 90]
})

# `math` 列总和。
print(df['math'].sum())
# `english` 列最大值。
print(df['english'].max())
# `physics` 列最小值。
print(df['physics'].min())
# `math` 列平均值。
print(df['math'].mean())
# `english` 列中位数。
print(df['english'].median())
# `physics` 列众数。
print(df['physics'].mode().values.tolist())
# `math` 列标准差、方差、0.25 分
print(df['math']. std(), df['math'].var(), df['math'].quantile(0.25))
````

### 题目12：重复值处理

已知：

```python
df = pd.DataFrame({
    'name': ['tom', 'tom', 'jack', 'alice', 'bob', 'allen'],
    'age': [15, 15, 15, 20, 26, 30],
    'score': [60.5, 60.5, 80, 30.6, 70, 83.5]
})
```

完成以下要求：

1. 查看所有完全重复记录的出现次数。
2. 去除重复行。
3. 使用 `duplicated(subset=['age'])` 判断 `age` 列是否重复。

````python
# 创建DataFrame
df = pd.DataFrame({
    'name': ['tom', 'tom', 'jack', 'alice', 'bob', 'allen'],
    'age': [15, 15, 15, 20, 26, 30],
    'score': [60.5, 60.5, 80, 30.6, 70, 83.5]
})
# 查看所有完全重复记录出现的次数
print(df.duplicated().sum())
# 去除重复行
df = df.drop_duplicates()
# 使用 duplicated(subset=['age']) 判断 age 列是否重复
print(df.duplicated(subset=['age']))
````

## 题目13：抽样、替换与累计计算

沿用题目12中的 `df`，完成以下要求：

1. 随机抽取 2 行。
2. 将 `age` 中的 `15` 替换成 `30`。
3. 对数值列分别计算累计和 `cumsum()`。
4. 对数值列按列方向计算累计最小值 `cummin(axis=0)`。

## 题目14：排序与 TopN

已知：

```python
df = pd.DataFrame({
    'name': ['tom', 'tom', 'jack', 'alice', 'bob', 'allen'],
    'age': [15, 15, 15, 20, 26, 30],
    'score': [60.5, 60.5, 80, 30.6, 70, 80]
}, index=[1, 2, 3, 4, 5, 6])
```

完成以下要求：

1. 按索引降序排序。
2. 按 `score` 升序排序。
3. 按 `score` 升序、`age` 降序联合排序。
4. 找出 `score` 最大的前 2 条记录。
5. 找出 `score` 最小的前 2 条记录。

````python
# 创建DataFrame
df = pd.DataFrame({
    'name': ['tom', 'tom', 'jack', 'alice', 'bob', 'allen'],
    'age': [15, 15, 15, 20, 26, 30],
    'score': [60.5, 60.5, 80, 30.6, 70, 80]
}, index=[1, 2, 3, 4, 5, 6])
# 按索引降序排序
print(df.sort_index(ascending=False))
# 按score升序排序
print(df.sort_values('score', ascending=True))
# 按score升序、age降序排序
print(df.sort_values(['score', 'age'], ascending=[True, False]))
# 找出score最大的前2条记录
print(df.nlargest(2, columns='score'))
# 找出score最小的前2条记录
print(df.nsmallest(2, columns='score'))
````
