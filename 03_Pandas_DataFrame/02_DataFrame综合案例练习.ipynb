## 案例1：学生成绩分析

已知某班级学生成绩数据如下：

```python
data = {
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '数学': [85, 92, 78, 88, 95],
    '英语': [90, 88, 85, 92, 80],
    '物理': [75, 80, 88, 85, 90]
}
scores = pd.DataFrame(data)
```

完成以下要求：

1. 计算每位学生的总分和平均分。
2. 找出数学成绩大于 90 分或英语成绩大于 85 分的学生。
3. 按总分从高到低排序，并输出前 3 名学生。
4. 找出三门课中单科最高分对应的学生和科目。
5. 增加一列 `是否偏科`：若某学生三科最高分与最低分之差大于等于 12，则记为 `是`，否则记为 `否`。

````python
# 创建学生分数序列scores
data = {
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '数学': [85, 92, 78, 88, 95],
    '英语': [90, 88, 85, 92, 80],
    '物理': [75, 80, 88, 85, 90]
}
scores = pd.DataFrame(data)
# 1. 计算每位学生的总分和平均分
scores['总分'] = scores[['数学', '英语', '物理']].sum(axis=1)
scores['平均分'] = scores[['数学', '英语', '物理']].mean(axis=1)
print(scores)
# 2. 找出数学成绩大于 90 分或英语成绩大于 85 分的学生
print(scores[(scores['数学'] > 90) | (scores['英语'] > 85)])
# 3. 按总分从高到低排序，并输出前 3 名学生
print(scores.nlargest(3, columns='总分'))
# 4. 找出三门课中单科最高分对应的学生和科目
print("数学最高分:", scores.nlargest(1, columns='数学'))
print("英语最高分:", scores.nlargest(1, columns='英语'))
print("物理最高分:", scores.nlargest(1, columns='物理'))
# 5. 增加一列 `是否偏科`：若某学生三科最高分与最低分之差大于等于 12，则记为 `是`，否则记为 `否`。
# 每个学生的最高分学科
max_scores = scores[['数学', '英语', '物理']].max(axis=1)
# 每个学生的最低分学科
min_scores = scores[['数学', '英语', '物理']].min(axis=1)
# np.where我忘记了，GPT提醒的
scores['是否偏科'] = np.where(max_scores - min_scores >= 12, '是', '否')
print(scores)
````

## 案例2：销售数据分析

已知某公司销售数据如下：

```python
data = {
    '产品名称': ['A', 'B', 'C', 'D', 'E'],
    '单价': [100, 150, 200, 120, 180],
    '销量': [50, 30, 20, 40, 25]
}
sales_df = pd.DataFrame(data)
```

完成以下要求：

1. 计算每种产品的总销售额。
2. 找出销售额最高的产品。
3. 按销售额从高到低排序，并输出所有产品信息。
4. 增加一列 `销量等级`：销量大于等于 40 为 `高`，20 到 39 为 `中`，否则为 `低`。
5. 统计总销售额、平均单价和平均销量。

````python
# 创建销售数据df
data = {
    '产品名称': ['A', 'B', 'C', 'D', 'E'],
    '单价': [100, 150, 200, 120, 180],
    '销量': [50, 30, 20, 40, 25]
}
sales_df = pd.DataFrame(data)
# 1. 计算每种产品的总销售额
sales_df['总销售额'] = sales_df['单价'] * sales_df['销量']
# 2. 找出销售额最高的产品
print(sales_df.nlargest(1, columns='总销售额'))
# 3. 按销售额从高到低排序，并输出所有产品信息
sales_df = sales_df.sort_values('总销售额', ascending=False)
print(sales_df)
# 4. 增加一列 `销量等级`：销量大于等于 40 为 `高`，20 到 39 为 `中`，否则为 `低`。
# 多条件判断:np.select 单条件判断:np.where
cdlist = [sales_df['销量'] >= 40,
          (sales_df['销量'] >= 20) & (sales_df['销量'] <= 39),
          sales_df['销量']< 20 ]
choice_list = ['高', '中', '低']
sales_df['销量等级'] = np.select(cdlist, choice_list)
print(sales_df)
# 5. 统计总销售额、平均单价和平均销量
print("总销售额:", sales_df['总销售额'].sum())
print("平均单价:", sales_df['单价'].mean())
print("平均销量:", sales_df['销量'].mean())
````

## 案例3：电商用户行为分析

已知电商平台用户行为数据如下：

```python
data = {
    '用户ID': [101, 102, 103, 104, 105],
    '用户名': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'],
    '商品单价': [1200, 300, 800, 150, 200],
    '购买数量': [1, 3, 2, 5, 4]
}
df = pd.DataFrame(data)
```

完成以下要求：

1. 计算每位用户的总消费金额。
2. 找出消费金额最高的用户，并输出其所有信息。
3. 计算所有用户的平均消费金额，保留 2 位小数。
4. 统计电子产品的总购买数量。
5. 按 `商品类别` 统计每类商品的总消费金额和总购买数量。
6. 找出购买数量大于平均购买数量的用户。

````python
# 创建电商平台用户行为数据df
data = {
    '用户ID': [101, 102, 103, 104, 105],
    '用户名': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'],
    '商品单价': [1200, 300, 800, 150, 200],
    '购买数量': [1, 3, 2, 5, 4]
}
df = pd.DataFrame(data)
# 1. 计算每位用户的总消费金额
df['总消费金额'] = df['商品单价'] * df['购买数量']
# 2. 找到消费金额最高的用户，并输出其所有信息
print(df.nlargest(1, '总消费金额'))
# 3. 统计所有用户的平均消费金额，保留 2 位小数
print(f"所有用户的平均消费金额:{df['总消费金额'].mean():.2f}")
# 4. 统计电子产品的总购买数量
print("电子产品的总购买数量:", df[df['商品类别'] == '电子产品']['购买数量'].sum())
# 5. 按 `商品类别` 统计每类商品的总消费金额和总购买数量
print("按 `商品类别` 统计每类商品的总消费金额和总购买数量")
print(df.groupby('商品类别')[['总消费金额', '购买数量']].sum())
# 6. 找出购买数量大于平均购买数量的用户
aver_num = df['购买数量'].mean()
print("购买数量大于平均购买数量的用户")
print(df[df['购买数量'] > aver_num])
````

## 案例4：宽表转长表与长表转宽表

已知学生成绩宽表如下：

```python
data = {
    'ID': [1, 2, 3],
    'name': ['张三', '李四', '王五'],
    'Math': [90, 85, 88],
    'English': [88, 92, 84],
    'Science': [95, 89, 91]
}
df = pd.DataFrame(data)
```

完成以下要求：

1. 将宽表转成长表，列名分别命名为 `科目` 和 `分数`。
2. 按 `name` 和 `科目` 排序。
3. 再将长表还原成宽表。
4. 对比宽表和还原后的表，观察索引和列结构有什么差异。

````python
# 创建学生成绩宽表
data = {
    'ID': [1, 2, 3],
    'name': ['张三', '李四', '王五'],
    'Math': [90, 85, 88],
    'English': [88, 92, 84],
    'Science': [95, 89, 91]
}
df = pd.DataFrame(data)
print(df)
# 将宽表转成长表（即多列同类字段变成类别类+数值类）
long_df = df.melt(id_vars=['ID','name'], value_vars=['Math','English','Science'],
        var_name='科目', value_name='分数')
# 按 `name` 和 `科目` 排序
print(long_df.sort_values(['name', '科目']))
# 将长表还原成宽表
# GPT:若想让还原后的宽表更接近原表，添加reset.index()方法
short_df = long_df.pivot(index=['ID','name'],columns='科目',values='分数').reset_index()
print(short_df)
# 原来的宽表索引是隐藏索引0,1,2...,还原后的表索引变成了（ID, name）
# 原来的表列结构是(ID,name,Math,English,Science)，还原后的表结构是(Math,English,Science)

````
